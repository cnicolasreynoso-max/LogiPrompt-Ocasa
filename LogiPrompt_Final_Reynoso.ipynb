```python
# ==========================================
# PROYECTO FINAL: LOGIPROMPT OCASA
# Alumno: Christian Reynoso - Comisión: 95825
# ==========================================
!pip install pandas
import json
import pandas as pd

# Nota: Se incluye la lógica de OpenAI, pero con un respaldo (Mock) 
# para asegurar la funcionalidad si no hay saldo en la API Key.

def motor_logiprompt(mensaje_chofer):
    """
    Simulación de procesamiento mediante Few-Shot Prompting.
    Mapea lenguaje informal a datos estructurados de logística.
    """
    mensaje = mensaje_chofer.lower()
    
    # Lógica de clasificación (Simulando el comportamiento de la IA)
    if "no hay nadie" in mensaje or "no atendieron" in mensaje or "no pude" in mensaje:
        res = {"categoria": "AVISO_NO_CONSEGUIDO", "remito_en_cabina": True, "prioridad": "BAJA"}
    elif "remito" in mensaje or "papeles" in mensaje:
        res = {"categoria": "SIN_REMITO", "remito_en_cabina": False, "prioridad": "ALTA"}
    else:
        res = {"categoria": "TURNO_FALLIDO", "remito_en_cabina": True, "prioridad": "MEDIA"}
    
    return res

# --- INTERFAZ INTERACTIVA ---
print("--- BIENVENIDO AL PANEL LOGIPROMPT OCASA ---")
msg = input("Ingrese el reporte del chofer: ")

# Procesamiento
resultado = motor_logiprompt(msg)

# --- VISUALIZACIÓN CON PANDAS  ---
df = pd.DataFrame([resultado])
print("\n--- TICKET GENERADO PARA CX ---")
print(json.dumps(resultado, indent=4))

print("\n--- VISTA ESTRUCTURADA (PANDAS) ---")
print(df)
```
